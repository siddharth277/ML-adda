```python
from paddleocr import PaddleOCR

#from paddleocr.tools.infer.utility import draw_ocr

from matplotlib import pyplot as plt
import cv2
```

# Initialization of Model and Detection

```python
ocr_model = PaddleOCR(use_angle_cls=True, lang='en', enable_mkldnn=False)
```

```python
import os
img_path = os.path.join('.', 'medicine.jpg')
```

```python
result = ocr_model.ocr(img_path)
```

```python

res = result[0]


boxes = res['dt_polys']
texts = res['rec_texts']
scores = res['rec_scores']
```

```python
# Specifying font path for draw_ocr method
font_path = os.path.join('PaddleOCR', 'doc', 'fonts', 'latin.ttf')
```

```python

img = cv2.imread(img_path) 


img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) 
```

```python
import numpy as np
annotated_img = img.copy() 

for box, text, score in zip(boxes, texts, scores):
    
    box = np.array(box).astype(int)
    
    cv2.polylines(annotated_img, [box], isClosed=True, color=(0, 255, 0), thickness=2)
    
    cv2.putText(annotated_img, text, (box[0][0], box[0][1] - 5), 
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 0, 0), 2)

plt.figure(figsize=(15, 15))
plt.imshow(annotated_img)
plt.axis('off') 
plt.show()
```
